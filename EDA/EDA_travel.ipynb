```python
plt.figure(figsize=(15,6))
sns.countplot(x="ProductPitched",data=df,palette="viridis")
plt.title("univariate anlysis")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x="ProdTaken",data=df,palette="viridis")
plt.title("univariate anlysis")

plt.show()
```

```python
for cat in cat_columns:
    plt.figure(figsize=(10,6))
    sns.countplot(x=cat,data=df,palette="viridis")
    plt.title(f"univariate analysis - {cat}")

    plt.show()
```

```python
# univariate analysis for num volumns
for num in num_columns:
    plt.figure(figsize=(10,6))
    sns.histplot(x=num,data = df,kde=True, bins = 20)
    plt.title(f"univariate analysis - {num}")

    plt.show()
```
