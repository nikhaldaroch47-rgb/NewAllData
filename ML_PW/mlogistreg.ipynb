```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns

import warnings 
warnings.filterwarnings("ignore")
```

```python
from sklearn.datasets import make_classification 
```

```python
x,y = make_classification(n_samples=1000,n_features=10,n_classes=3,n_redundant=5,n_informative=5,random_state=1)
```

```python
x
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
# ovr (one versus rest)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.30,random_state=1)
```

```python
model = LogisticRegression(multi_class="ovr",solver="lbfgs")
```

```python
model
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix , accuracy_score
```

```python
confusion_matrix(y_test , y_pred)
```

```python
accuracy_score(y_test , y_pred)
```

```python
from sklearn.datasets import load_iris
```

```python
data = load_iris()
```

```python
print(data.DESCR)
```

```python
data.target
```

```python
data.feature_names
```

```python
df = pd.DataFrame(data.data,columns=data.feature_names)
```

```python
df["target"] = data.target
```

```python
df
```

```python
x1 = df.iloc[:,:-1]
```

```python
y1 = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x1_train , x1_test , y1_train , y1_test = train_test_split(x1,y1,test_size=0.30,random_state=1)
```

```python
model2 = LogisticRegression(max_iter=300)   # we are not passing anything by default it will dectect it is multiclass classification
```

```python
model2
```

```python
model2.fit(x1_train,y1_train)
```

```python
y_pred = model2.predict(x1_test)
```

```python
confusion_matrix(y1_test,y_pred)
```

```python
accuracy_score(y1_test,y_pred)
```

```python
model2.predict_proba(x1_test) # tools is roc curves 
```
