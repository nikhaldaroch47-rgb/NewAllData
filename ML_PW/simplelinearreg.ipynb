```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
import numpy as np
```

```python
# here we are loading dataset from scitlearn
from sklearn.datasets import load_diabetes
```

```python
diabetics = load_diabetes()
```

```python
diabetics
```

```python
# problem statement -> to predict disease progression one year after baseline provided 10 features
# (in one year how much sugar level change)
```

```python
diabetics.data
```

```python
diabetics.target
```

```python
diabetics.feature_names
```

```python
data = pd.DataFrame(diabetics.data,columns=diabetics.feature_names)
```

```python
data["target"] = diabetics.target
```

```python
data
```

```python
x = data[["bmi"]]   # independent variable 
y = data["target"]  # dependent variable 
```

```python
from sklearn.model_selection import train_test_split  # sklearn is a liberary and model_selection is a module in it  inside that their is a function
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state=1) # random state means that value will not change after every excetution any number of times
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr  # this type of color means model is not trained yet  
```

```python
# training of model 
lr.fit(x_train,y_train)
# now see the color of object
# at the backend it will use gradient desent to give optimal cofficient
```

```python
y_pred = lr.predict(x_test)
```

```python
# visulazitation 
plt.scatter(x_test,y_test,color="black",label="actual")
plt.plot(x_test,y_pred,color="blue",label="best fit line of liner reg")
```

```python
# model evaluation 
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(y_test,y_pred)
```
