```python
from xgboost import XGBRegressor   # xg boost is a seperate liberary 
```

```python
from sklearn.datasets import make_regression
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
x , y = make_regression(n_samples = 1000 , n_features=20,n_informative=10,noise=0.1,random_state=1)
```

```python
x
```

```python
y
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
re = XGBRegressor()
```

```python
re   # since xg is a seperate liberary it have so many parameters to tune 
```

```python
re.fit(x_train , y_train)
```

```python
y_pred = re.predict(x_test)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
params = {
    'n_estimators':[100,200],
    'learning_rate':[0.01,0.1],
    'max_depth':[3,5],
    }
```

```python
reg = XGBRegressor()
```

```python
grid = GridSearchCV(estimator=reg , param_grid=params , cv=5)
```

```python
grid.fit(x_train , y_train )
```

```python
grid.best_params_
```

```python
final = grid.best_estimator_
```

```python
final.predict(x_test)
```
