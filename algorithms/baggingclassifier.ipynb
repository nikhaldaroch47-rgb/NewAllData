```python
# we use label encoder 
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
ans = le.fit_transform(df["time"])
```

```python
df["time"] = ans 
```

```python
x = df.drop("time",axis=1)
```

```python
y = df["time"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y ,test_size=0.2,random_state=1)
```

```python
from sklearn.impute import SimpleImputer # for missing value treatment 
from sklearn.preprocessing import OneHotEncoder # for encoding
from sklearn.preprocessing import StandardScaler  # for scaling
```

```python
from sklearn.pipeline import Pipeline  # a sequesnce of data transformer 
from sklearn.compose import ColumnTransformer # group all the pipeline steps to all the columns
```

```python
df.sample(1)
```

```python
num_cols = ["total_bill","tip","size"]
cat_cols = ["sex","smoker","day"]
```

```python
num_pipeline = Pipeline(steps=[("imputation",SimpleImputer(strategy="median")),("scaling",StandardScaler())])
```

```python
cat_pipeline = Pipeline(steps=[("imputation",SimpleImputer(strategy="most_frequent")),("encoding",OneHotEncoder())])
```

```python
preprosser = ColumnTransformer([("num_pipeline",num_pipeline,num_cols),("cat_pipeline",cat_pipeline,cat_cols)])
```

```python
x_train = preprosser.fit_transform(x_train)
x_test = preprosser.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
```

```python
model = {"logisticreg":LogisticRegression(),"supportvector":SVC(),"decisiontree":DecisionTreeClassifier()}
```

```python
for i in range(len(model)):
    print(list(model.values())[i])
```

```python
from sklearn.metrics import accuracy_score

def model_train_evaluation(x_train,x_test,y_train,y_test,model):
    evaluation = {}
    for i in range(len(model)):
        models = list(model.values())[i]
        models.fit(x_train,y_train)
        y_pred = models.predict(x_test)
        model_score = accuracy_score(y_test,y_pred)
        evaluation[list(model.keys())[i]] = model_score
    return evaluation 
```

```python
model_train_evaluation(x_train,x_test,y_train,y_test,model)
```
