```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier

import warnings 
warnings.filterwarnings("ignore")
```

```python
x , y = make_classification(n_samples=1000,n_features=20,n_classes=2,random_state=1)  #random state means to reproduce result only once
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.30,random_state=1)
```

```python
adb = AdaBoostClassifier()
```

```python
adb
```

```python
adb.fit(x_train,y_train)
```

```python
y_pred = adb.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,recall_score,confusion_matrix
```

```python
accuracy_score(y_pred,y_test)
```

```python
confusion_matrix(y_pred , y_test )
```

```python
from sklearn.model_selection import  GridSearchCV
```

```python
param_grid = {"n_estimators" : [50,100,200],"learning_rate" : [0.01,0.1,1.0,1.5,2.0],"algorithm" : ["SAMME"]}
```

```python
ADDA = AdaBoostClassifier()
clf = GridSearchCV(estimator=ADDA,param_grid=param_grid,cv=5,verbose=2)
```

```python
clf.fit(x_train,y_train)
```

```python
clf.best_params_
```

```python
model = clf.best_estimator_
```

```python
y_pred = model.predict(x_test)
```

```python
accuracy_score(y_pred , y_test)
```

```python
from sklearn.datasets import make_regression
```

```python
from sklearn.ensemble import AdaBoostRegressor
```

```python
x,y = make_regression(n_samples=1000,n_features=2,noise=10,random_state=1)
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.20,random_state=1)
```

```python
adr = AdaBoostRegressor()
```

```python
adr
```

```python
adr.fit(x_train , y_train)
```

```python
y_pred = adr.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error
```

```python
mean_squared_error(y_test , y_pred )
```

```python
r2_score(y_pred , y_test )
```
