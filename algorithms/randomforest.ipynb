```python
df = sns.load_dataset("tips")
```

```python
df.time.unique()
```

```python
x = df.drop("time",axis=1)
x
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df["time"] = le.fit_transform(df.time)
```

```python
y = df.time
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.30,random_state=1)
```

```python
from sklearn.impute import SimpleImputer # for missing values 
from sklearn.preprocessing import OneHotEncoder # encoding
from sklearn.preprocessing import StandardScaler # for scaling
```

```python
from sklearn.pipeline import Pipeline  # a sequesnce of data transformer 
from sklearn.compose import ColumnTransformer # group all the pipeline steps to all the columns
```

```python
df.time.unique()
```

```python
df.time.unique()
```

```python
num_cols = ["total_bill","tip","size"]
cat_cols = ["sex","smoker","day"]
```

```python
num_pipeline = Pipeline(steps=[("imputation",SimpleImputer(strategy="median")),("scaling",StandardScaler())])
```

```python
cat_pipeline = Pipeline(steps=[("imputation",SimpleImputer(strategy="most_frequent")),("encoding",OneHotEncoder())])
```

```python
preprosser = ColumnTransformer([("num_pipeline",num_pipeline,num_cols),("cat_pipeline",cat_pipeline,cat_cols)])
```

```python
x_train = preprosser.fit_transform(x_train)
x_test = preprosser.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
```

```python
models = {"decision_tree":DecisionTreeRegressor(),"linear":LinearRegression()
,"svr":SVR(),"random_forest":RandomForestRegressor()}
```

```python
models 
```

```python
for i in range(len(models)):
    print(list(models.values())[i])
```

```python
from sklearn.metrics import accuracy_score

def model_train_eval(x_train,x_test,y_train,y_test,models):
    evaluation = {}
    for i in range(len(models)):
        model = list(models.values())[i]
        model.fit(x_train,y_train)
        y_pred = model.predict(x_test)
        model_score = accuracy_score(y_pred , y_test )
        evaluation[list(models.keys())[i]] = model_score
    return evaluation
```

```python
model_train_eval(x_train,x_test,y_train,y_test,models)
```
