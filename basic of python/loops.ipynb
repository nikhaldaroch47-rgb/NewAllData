```python
# while loop ->>> the code executed untill the condition is not hit
# syntax
#''' while condition 
 #             execute a block of code'''
```

```python
n = 7 
i = 2
while i < n:
    print(i)
    i+=2
```

```python
count = 6
while count > 0:
    print(count)
    count-=2
```

```python
n = 7 
i = 1
while i < n:
    print(i)
    i+=1 
    if i == 4:
        break  # break wiil completely break the loop and come out of the loop
```

```python
# to print even numbers using while loop
n = 20
i = 0
while i < n:
    i+=1
    if i % 2 != 0:
        continue  #  
    print(i , end = " ")
```

```python
n = 20  # odd numbers 
i = 0
while i < n:
    i+=1
    if i % 2 == 0:
        continue  #  
    print(i , end = " ")
```

```python
# for loop ->> will only iterate the sequence of elements 
a = "gurnooor"
for i in a:
    print(i , end="")
```

```python
l =[1,2,3,"nikhil","sukh"]
for i in l:
    if i=="nikhil":
        break  # same with continue but only for particular iteration
    print(i,end= " ")
```

```python
# range(0 , 10)  ->> it will give the value from 0 to 9
for i in range(0,10):
    print(i,end=", ")  # end will assign all the elements in horizontal form
```

```python
row = 1
while row <=7:
    colmn=1
    while colmn<=row:
        print("*",end=" ")
        colmn+=1
        
    print()
    row+=1     
```

```python
for i in range(4):
    for j in range(4):
        print("*" , end = " ")
    print()    
```

```python
# pattern of square...
row = 0
while row<4:
    column=0
    while column<4:
        print("*", end = " ")
        column+=1  
    row+=1    
    print()    
        
```
