```python

def letter(l):
    lower_count=0
    upper_count=0
    for i in l:
        if i.isupper:
            upper_count+=1
        elif i.islower:
            lower_count+=1
    return upper_count,lower_count
```

```python
count = letter("Nikhil")
print(count)
```
