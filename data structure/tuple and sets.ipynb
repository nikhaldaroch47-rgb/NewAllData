```python
person1 = {"reading", "bhangra","coding"}
person2 = {"bhangra", "traveling", "hiking"}
print(person1)
print(person2)
```

```python
# unoin ->> combing all the elements 
person1 | person2
```

```python
# intersection - >> common in both the sets
person1 &person2
```

```python
# difference ->> show only element present in first set not in sec
person1 - person2
```

```python
# symmetric ->> not give the common part
person1 ^ person2
```
