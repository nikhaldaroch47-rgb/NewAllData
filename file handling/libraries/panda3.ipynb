```python
data = {"key1":[1,2,3,4,5],
"key2":[6,7,8,9,10],"key3":[11,12,13,14,15]}
d1 = pd.DataFrame(data)
d1
```

```python
d2 = pd.DataFrame({"key1":[1,2,6,5,8,8],
"key4":[6,7,8,9,0,11],"key5":[1,2,3,4,5,10]})
d2
```

```python
pd.merge(d1,d2,how="inner")  # will merge two dataframes based on their common elements 
```

```python
pd.merge(d1,d2,how="left")  # left data frame will be printed and based on left right elements will be printed which are comman
```

```python
pd.merge(d1,d2,how="right")  # similar to right merge
```

```python
pd.merge(d1,d2,how="outer")   # all the comman elemnets of the common column from both data frame
```

```python
pd.merge(d1,d2,how="cross") # all permutation will be possible
```

```python
pd.merge(d1,d2,how="right",right_on="key4",left_on="key2")  # when the columns are not same 
```
