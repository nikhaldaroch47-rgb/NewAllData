```python
class Father:
    def father_prop(self):
        print("this is father proprty")
class son(Father):
    def job(self):
        print("this is son property")        
```

```python
obj = son()
obj.father_prop()
```

```python
obj1 = Father()
obj1.father_prop()
obj1.job() #--->> this will give an error because father will only have there own property
```

```python
class Fruit:
    def fruit_info(self):
        print("this is parent")
class apple(Fruit):
    def apple_info(self):
        print("this is child class")        
```

```python
a = Fruit()
a.fruit_info()
a.apple_info() # --->>> this will give an error because fruit have there own property
```

```python
b = apple()
b.apple_info()
b.fruit_info()
```

```python
#  METHOD OVERRIDING -->> means rewirting and redefining the parent class in derived class

class Fruit:
    def fruit_info(self):
        print("this is parent")
class apple(Fruit):
    def fruit_info(self):
        print("this is parent class")
    def apple_info(self):
        print("this is child class")  
```

```python
a = Fruit()
a.fruit_info()
```

```python
b = apple()
b.fruit_info()
```

```python
# MULTI-LEVEL-INHERETANCE --->> child class will inheret the property level-wise
class Grandfather:
    def gf_info(self):
        print("this is grand father property")
class Father(Grandfather):
    def father_info(self):   
        print("i am ur father and i have both my and gf property")
class Son(Father):
    def son_info(self):
        print("i have both father and grandfather property")             
```

```python
s1 = Grandfather()
s1.gf_info()
```

```python
s2 = Son()
s2.gf_info()
s2.son_info()
s2.father_info()
```

```python
# MULTIPLE INHERITANCE 
class Mother():
    def method1(self):
        print("meth of class mother")
class Father():
    def method2(self):
        print("meth of class father")
class Child(Mother,Father):
    def method(self):
        print("meth of child")
```

```python
c1 = Child()
c1.method2()
c1.method()
c1.method1()
```

```python
# DIAMOND PROBLEM -> IT OCCURS WHEN INHERETS 2 OR MORE THAN 2 CLASSES
# TO REMOVE DIAMOND PROB PYTHON USES METHOD RESOLUTION ORDER ( MRO)


# MEANING THAT THE CLASS IHERETED FIRST IN DRIVED CLASS THAT METH WILL BE CALLED

class Mother():
    def method1(self):
        print("meth of class mother")
class Father():
    def method1(self):
        print("meth of class father")
class Child(Mother,Father):
    def method(self):
        print("meth of child")
```

```python
a1 = Child()
a1.method1()
a1.method1()
```

```python
# diamond structure
class A():
    def method(self):
        print("meth of class A")
class B(A):
    def method(self):
        print("method of class B")
class C(A):
    def method(self):
        print("meth of classs C")           
class D(B,C):
    pass             
```

```python
d1 = D()
d1.method()
```

```python
# HIRERACHY INHERATNCE -->> WHEN ONE PARENT HAVE MULTIPLE CHILDS
class Vehicle():
    def info(self):
        print("vehicle info")
class Car(Vehicle):
    def car_info(self,name):
        print("car info is ",name)
class Truck(Vehicle):
    def truck_info(self,name):
        print("truck info is ",name)               
```

```python
s1 = Car()
s1.info()
s1.car_info("buggati")
s2 = Truck()
s2.truck_info("camper")
```
