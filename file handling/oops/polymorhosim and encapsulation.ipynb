```python
# polymorphism in oops takes two ways
# method overloading 
# method overriding
```

```python
# method overloading

class Student:
    def student(self):
        print("welcome student")
    def student(self,name =""):
        print("welcome student",name)
    def student(self,name ="",course =""):
        print("welcome student",name,course)        
```

```python
obj = Student()
obj.student()
obj.student("nikhil")
obj.student("student","data science")
```

```python
# method overloading  --->> student method taking different forms ,, the last method overloads the pervious ones..

class Student:
    def student(self,name ="",course =""):
        print("welcome student",name,course) 
```

```python
obj = Student()
obj.student("nikhil","data science")   
```

```python
class Animal:
    def sound(self):
        print("animal sound")
class Cat(Animal):
    def sound(self):
        print("cat sound -> meows")        
```

```python
a = Animal() #--> animal class
a.sound()
b = Cat() # -->> cat class
b.sound()
```

```python
# accessing the public data member in the other method of class
class Student:
    def __init__(self,name,degree):
        self.name = name
        self.degree = degree

    def show(self):
        # accessing public data membber in this class  
        print("name = ",self.name,", degree =",self.degree)  
```

```python
class Student:
    def __init__(self,name,degree):
        self.name = name
        self.__degree = degree
    def show(self):
        print("name :",self.name,"degree :",self.__degree)

    def __private_method(self):
        print("this method is private") 

    def access__private_method(self):
        self.__private_method()    
```

```python
obj1 = Student("nikhil","btech cse")
obj1.access__private_method()
```

```python
class Car:
    def __init__(self,year,speed,model):
        self.__year = year
        self.__speed = speed
        self.__model = model
    def set_speed(self,speed):
        self.__speed = 0 if speed<0 else speed
    def get_speed(self):
        return self.__speed       
```

```python
c1 = Car("2022","80","fortuner")
c1.set_speed(50)

c1.get_speed()
```

```python
# another use case
class Bank:
    def __init__(self,balance):
        self.__balance = balance
    def deposit(self,amount):
        self.__balance = self.__balance + amount
    def withdraw(self,amount):
        if self.__balance > amount:
            self.__balance = self.__balance - amount
            return True
        else:           
            return False
    def get_balance(self):
        return self.__balance         
```

```python
b = Bank(30000)
b.deposit(30000)
b.withdraw(1000)
b.get_balance()
```
