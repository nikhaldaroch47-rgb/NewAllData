```python
def fib_func(n):
    if n<=0:
        return []
    elif n==1:
        return [0]
    else:
        fib = [0,1]
        for i in range(2,n):
            next_fib = fib[i-1]+fib[i-2]
            fib.append(next_fib)
        return fib
fib_func(10)                    
```

```python
def fact_fun(n):
    fact = 1
    for i in range(1,n+1):
        fact*=i
    return fact
fact_fun(5)               
```

```python
def reverse_str(s):
    str1 = ""
    for i in s:
        str1 = i+ str1  # first in at the last and last at the first
    return str1
reverse_str("nikhil")        
```

```python
# deep copy - deep copy creates a new compund object before inserting copies of the item found in the original
import copy
a = [[1,2,3,4],[5,6,7,8,9]]
  
b = copy.deepcopy(a)
b[0][0]=10
b 
```

```python
# shallow copy - shalllow copyy creates a new object but retain refference to the object contained in original
import copy
a =[[1,2,3,4],[5,6,7,8]]
b = copy.copy(a)
b[0][0]=99
b 
```
