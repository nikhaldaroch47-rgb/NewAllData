```python
# write a code input as list and in list only numeric value is there
def numeric(a):
    n = []
    for i in a:
        if type(i)==int or type(i)==float:
            n.append(i)
    return n                
```

```python
numeric([1,2,1.2,True,3+5j,5])
```

```python
# [1,2,3,4.4,"nikhil",[100,1001,"meet"]]
def numeric(a):
    n = []
    for i in a:
        if type(i)==list:
            for j in i:
                if type(j)==int or type(j)==float:
                    n.append(j)
        else:
            if type(i)==int or type(i)==float:
                n.append(i)
        
    return n                          
```

```python
numeric([1,2,3,4.4,"nikhil",[100,1001,"meet"]])
```

```python
# varible len argument 
def sum1(*args):    # *args ->> as many inputs we want
    s = 0
    for i in args:
        s+=i
    return s        
```

```python
sum1(1,2,3,4,5,6,7)
```

```python
# write a function to return a list
def return_list(*args):   # **args is a keyword argument
    l = []
    for i in args:
        if type(i)==list:
            l.append(i)
    return l        
```

```python
return_list(1,2,3,[12,3,4,4,5])
```
