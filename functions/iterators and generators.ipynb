```python
# generator function 
# regular function 
def square_fn(n):
    l = []
    for i in n:
        l.append(i**2)
    return l 
square_fn([1,2,3,4,5,6,7,8])       
```

```python
# can we have something to generate result one by one
# yield ->> in general function we use return statement BUT in generator we use yield..
def square_generator(n):
    for i in range(n):
        yield i**2
        
```

```python
gen = square_generator(10)   # yield function will create a generator/iterator object
gen
```

```python
next(gen)
```

```python
def fib(n):
    a = 0
    b = 1
    for i in range(n):
        yield a
        a,b = b,a+b
fib(10)        
```

```python
f = fib(10)
f
```

```python
next(f)
```
