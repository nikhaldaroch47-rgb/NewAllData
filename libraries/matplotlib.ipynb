```python
# line chart ->> it a type of visulization in python when we have to the trend and want to comapre two values 

x = [1,2,3,4,5]  # x - axis no of days 
y = [5,6,4,8,1]  # y - axis share 
plt.plot(x,y,color="red",marker="o",linestyle="--")
plt.grid()
```

```python
x = [1,2,3,4,5]
y1 = [2,3,1,5,7]
y2 = [5,4,9,2,5]
plt.plot(x,y1,color="red",marker="o",linestyle="--",linewidth=3,label="y1 stoke share")
plt.plot(x,y2,color="blue",marker="o",linestyle="--",linewidth=3,label="y2 stoke share")
plt.grid()
plt.legend()
```

```python
fruits = ['Apples', 'Bananas', 'Cherries', 'Dates']
sales = [400, 350, 300, 450]
plt.bar(x=fruits,height=sales)
plt.show()
```

```python
plt.barh(y=fruits,width=sales)
plt.show()
```

```python
plt.bar(x=fruits,height=sales,width=0.5,color=["red","green","blue","pink"])
plt.xlabel("fruits")
plt.ylabel("sales count")
plt.title("graph of sales vs fruits")
plt.show()
```

```python
# pie chart 

country=['India','Pakistan','Bangladesh','Nepal','Spain']
values=[200,180,230,100,120]

plt.pie(x = values,labels=country,autopct="%1.0f%%")
plt.show()
```

```python
x = np.linspace(0,10,100)
x
```

```python
plt.plot(x,np.sin(x), linestyle= "--")
```

```python
plt.figure()
plt.subplot(1,2,1)
plt.plot(x,np.sin(x))


plt.subplot(1,2,2)
plt.plot(x,np.cos(x))
```

```python
plt.Figure()  # -->> used to creat a page 
plt.subplot(2,2,1)  # is used create sub plot (2->no of rows,2->no of columns,make at first)
plt.plot(x,np.sin(x))  

plt.subplot(2,2,2)
plt.plot(x,np.cos(x))

plt.subplot(2,2,3)
plt.plot(x,np.tan(x))

plt.subplot(2,2,4)
plt.plot(x,np.sin(x+1))

plt.show
```

```python
plt.plot(x,np.sin(x-0))
plt.ylim(-1.5,1.5)
plt.xlim(1,15)
plt.show
```

```python
# stacked histogram -->> when want to comapre two histogram 

x = np.random.normal(0,5,5000)
x
```

```python
y = np.random.normal(2,5,5000)
y
```

```python
plt.hist([x,y],color=["r","b"],label=["dist1","dist2"])
plt.title("stacked histogram")
plt.xlabel("x axis")
plt.ylabel("frequency")

plt.show
```

```python
# bar chart -->> to see the distribution of categrocial variables 

x = ["a","b","c","d"]
y = [2,5,6,3]

plt.bar(x,y)
```

```python
# pie chart = used to see the percent distribution of data

x = ["python","java","c++","c"]
y =[95,35,70,80]
plt.pie(y,labels=x,autopct="%1.1f%%",shadow=True)
plt.show()
```

```python
# 3d plot

x = np.random.rand(20)
y = np.random.rand(20)
z = np.random.rand(20)
fig = plt.figure()
ax = fig.add_subplot(projection="3d")
ax.scatter(x,y,z,c='r')
```
