```python
tips = sns.load_dataset("tips")
tips
```

```python
tips.info()
```

```python
sns.catplot(x = "smoker",y="tip",data=tips) # catplot is for categorical data
```

```python
# some of the data points are overlapping so we have swarm
sns.catplot(x = "smoker",y="tip",data=tips,kind="swarm")  # we can aslo use box plot to see the outlayers in kind = box
```

```python
# in catplot their is a barplot means in categorical data
sns.barplot(x="day",y="total_bill",data=tips,color="green")  # how total billl is variying with day
```

```python
# joint plot
sns.jointplot(x=tips.total_bill,y=tips.tip)
```

```python
sns.jointplot(x=tips.total_bill,y=tips.tip,kind="kde")  # kde --> kernel density estimater mostly used in weather forecasting
```

```python
sns.pairplot(tips,palette="coolwarm")  # is used to draw different plots 
```
