```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
import pandas as pd 
df= pd.DataFrame({'result':['good','bad','good','bad','excellent','bad','excellent','good','bad']})
df
```

```python
df.result.value_counts()
```

```python
# fit , transform , and fit_transform 
le.fit(df.loc[:,'result'])
```

```python
ans = le.transform(df.loc[:,"result"])
```

```python
df.loc[:,"new"] = ans
```

```python
df.value_counts()
```

```python
# we can also do directly fit_transform 
encoder = LabelEncoder()
ans2 = encoder.fit_transform(df.loc[:,"result"])
```

```python
df.loc[:,"new1"] = ans2   # meaning of one hpt encoder 
df
```

```python
from sklearn.preprocessing import OneHotEncoder
onehot = OneHotEncoder()
```

```python
onehot.fit(df.loc[:,["result"]])
```

```python
answ = onehot.transform(df.loc[:,["result"]]).toarray()  # why this array is used 
```

```python
onehot.get_feature_names_out()
```

```python
df.loc[:,onehot.get_feature_names_out()] = answ
```
