```python
#data interpolation -->> estimating unkown values from known values 

# 3 types of data interpolation 
# linear , cubic , polynomial 
```

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4,5])
y = np.array([1,2,3,4,5])
```

```python
plt.scatter(x,y)
```

```python
# linear interpolation --> it is linear in nature 
x_new = np.linspace(1,5,10)
y_interp = np.interp(x_new,x,y)
plt.scatter(x_new,y_interp)
```

```python
 # cubic interpolation 

x = np.array([1,2,3,4,5])
y = np.array([1,8,27,64,125])
plt.scatter(x,y)
```

```python
# using new library 
from scipy.interpolate import interp1d
f = interp1d(x,y, kind="cubic")
```

```python
x_new = np.linspace(1,5,10)
y_interp = f(x_new)
```

```python
plt.scatter(x_new,y_interp)
```

```python
# polynomial interpolation 
x = np.array([1,2,3,4,5])
y = np.array([1,4,9,1,25])

p =np.polyfit(x,y,2)  # any ploy  curve who"s digree is 2
```

```python
x_new = np.linspace(x,y,2)
y_interp = np.polyval(p,x_new)
plt.scatter(x_new,y_interp)
```
