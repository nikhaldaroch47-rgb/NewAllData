```python
# scaling 
import pandas as pd 
import seaborn as sns 
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = sns.load_dataset("tips")
```

```python
sns.distplot(df["total_bill"])
```

```python
mean = np.mean(df["total_bill"])
std = np.std(df[("total_bill")])
```

```python
standardised_data = []
for i in list(df["total_bill"]):
    z_score = (i-mean/std)
    standardised_data.append(z_score)
```

```python
standardised_data
```

```python
# dosent change the distribution of data

from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler
```

```python
scaler.fit(df[["total_bill"]]) # will calculate  mu and sigma 
```

```python
scaler.transform(df[["total_bill"]])
```

```python
scaler.fit_transform(df[["total_bill"]])
```

```python
pd.DataFrame(scaler.fit_transform(df[["total_bill"]]),columns=["total_bill"])
```

```python
# min max scaler 
from sklearn.preprocessing import MinMaxScaler
min_max = MinMaxScaler()
min_max.fit_transform(df[["total_bill"]])
```
