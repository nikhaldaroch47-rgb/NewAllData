```python
import pandas as pd
import numpy as np 

np.random.seed(1)  # pseudo random number generator
```

```python
no_samples = 1000
no_class0_ratio = 0.9
no_class_0 = int(no_samples*no_class0_ratio)
no_class_1 = 100

no_class_0,no_class_1
```

```python
class_0 = {"feature0":np.random.normal(0,1,no_class_0),
"feature1":np.random.normal(0,1,no_class_0),
"target": [0]*no_class_0
}
df = pd.DataFrame(class_0)
df
```

```python
class_1 = {"feature0":np.random.normal(3,1,no_class_1),
"feature1":np.random.normal(3,1,no_class_1),
"target": [1]*no_class_1
}
df1 = pd.DataFrame(class_1)
```

```python
data = pd.concat([df,df1]).reset_index(drop=True)
data
```

```python
data.target.value_counts()
```

```python
# up sampling means to repeat the minority class data 
```

```python
data_majority = data[data.target == 0]
```

```python
data_minority = data[data.target == 1]
```

```python
# replace - if replace is true then -> it means data will be repeated 
from sklearn.utils import resample
data_minority_upsampled = resample(data_minority,replace=True,n_samples =len(data_majority),random_state=1)
```

```python
data_minority_upsampled.shape
```

```python
data_upsampled = pd.concat([data_majority,data_minority_upsampled])
data_upsampled.target.value_counts()
```

```python
# DOWN SAMPLING
```

```python
data_majority_down_sampling = resample(data_majority,replace=False,n_samples=len(data_minority),random_state=1)
```
