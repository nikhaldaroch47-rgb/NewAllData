```python
class Car:  # function/method of the car
    def accelerate(self):
        print("car is accelerating")
# self is represent the instance of the class using which we can access method/function of the class also the attribute/property of the class        
```

```python
C1 = Car()
C1.accelerate()
# self is not a python keyword but it is best practice to us it
```

```python
class Car:  # function/method of the car
    def accelerate(self):
        print("car is accelerating")
    def brake(self):
        print("car is stopping")    
```

```python
class listopration():
    l = [1,2,3,4,5,6,7,8,9]
    def extract_even(self,l):
        l1=[]
        for i in l:
            if i%2 == 0:
                l1.append(i)
        return l1   
    def extract_odd(self,l):
        l1 = []
        for i in l:
            if i%2 != 0:
                l1.append(i)
        return l1             
```
