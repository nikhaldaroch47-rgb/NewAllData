```python
# right angle triangle 

for i in range(1,5):   # oter loop is for rows 
    for j in range(1,i+1): # inner is for columns
        print("*",end=" ")
    print() # this is for next line
```

```python
# ulta right angle triangle
for i in range(5,0,-1):
    for j in range(1,i+1):
        print("*",end=" ")
    print()
```

```python

for i in range(1,5): # for rows 
    for j in range(5-i): # for spaces 
        print(" ",end=" ")
    for k in range(1,2*i): # for stars 
        print("*",end=" ")
    print()
```

```python
for i in range(5,0,-1):  # for rows 
    for j in range(5-i):    # for spaces 
        print(" ",end=" ")
    for k in range(1,2*i):    # for stars 
        print("*",end=" ")  
    print()  
```

```python
# hollow square
for i in range(1,6):
    for j in range(1,6):
        if i==1 or i==5 or j==1 or j==5:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```
